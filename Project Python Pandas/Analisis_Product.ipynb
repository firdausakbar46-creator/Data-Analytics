# Data Visualization

```python
# Plotting Bar Chart Area

ax = sns.countplot(x = 'Area', data= df_cleaned, hue='Area')

for bars in ax.containers:
    ax.bar_label(bars)
```

Jumlah data west dan east

```python
# Plotting Bar Chart data Region vs Score

item_score = df.groupby(['Region'], as_index= False)['Score'].mean().sort_values(by= 'Score', ascending= False)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = df_cleaned, x= 'Region', y= 'Score', hue= 'Region')
```

```python
ax= sns.countplot(x='Pareto', data= df_cleaned, hue= 'Pareto')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
month_score = df.groupby(['Month'], as_index=False)['Score'].mean().sort_values(by='Month', ascending=False)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = month_score, x = 'Month',y= 'Score', hue= 'Month')
```
